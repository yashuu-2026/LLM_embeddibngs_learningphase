Import Libraries

```python
import pandas as pd
import seaborn as sns
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error , mean_absolute_error , r2_score
```

Load Dataset

```python
df.head()
```

Finding the NULL values and dropping

```python
df = pd.read_csv("/content/CAR DETAILS FROM CAR DEKHO.csv")
print("Before dropping null values:\n")
print(df.isnull().sum())
df.dropna(inplace=True)
print("After dropping null values:\n")
print(df.isnull().sum())
```

Check duplicates and drop

```python
print("number of dublicated",df.duplicated().sum())
df.drop_duplicates(inplace=True)
print("number of duplicated after dropping:", df.duplicated().sum())
```

Apply label encoding on Fuel and Transmission columns

```python
FUEL = LabelEncoder()
df["fuel_encoded"] = FUEL.fit_transform(df["fuel"])
df.head(10)
```

```python
TRAN = LabelEncoder()
df["transmission_encoded"] = TRAN.fit_transform(df["transmission"])
df.head(10)
```

Separate X and Y

```python
X = df[["year","km_driven","fuel_encoded","transmission_encoded"]]
Y = df["selling_price"]
print("Feature shape:", X.shape)
print("Target shape:", Y.shape)
```

Split Train test(20%)

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
print("records for training:",len(X_train))
print("records for testing:", len(X_test))
```

Apply MinMax Scalar

```python
num_cols = ["year","km_driven"]
scaler = StandardScaler()
X_train[num_cols] = scaler.fit_transform(X_train[num_cols])
X_test[num_cols] = scaler.transform(X_test[num_cols])
print(X_train)
```

create linear regression model and fit with Xtrain and ytrain

```python
car_model = LinearRegression()
car_model.fit(X_train, Y_train)
```

Predicting selling prices and compare actual vs predicted prices

```python
y_pred = car_model.predict(X_test)
comparison_df = pd.DataFrame({'Actual Prices': Y_test, 'Predicted Prices': y_pred})
comparison_df.head()
```
